```python
!pip install yfinance pandas

import yfinance as yf
import pandas as pd

START = "2020-01-01"
END = "2026-01-01"

def download_stock(ticker):
    df = yf.download(
        ticker,
        start=START,
        end=END,
        auto_adjust=False
    )

    df = df.reset_index()

    df = df[
        ["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"]
    ]

    return df


jpm = download_stock("JPM")
bac = download_stock("BAC")
wfc = download_stock("WFC")



print(jpm.head())
print(jpm.shape)

jpm.to_csv("JPM.csv", index=False)
bac.to_csv("BAC.csv", index=False)
wfc.to_csv("WFC.csv", index=False)


```

```python
# Kiểm tra quality
for df, name in [(jpm,"JPM"), (bac,"BAC"), (wfc,"WFC")]:
    print(name)
    print(df.isna().sum())
```
