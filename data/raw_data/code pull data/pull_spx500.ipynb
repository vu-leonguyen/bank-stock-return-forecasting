```python
!pip install yfinance pandas

import yfinance as yf
import pandas as pd
import numpy as np

START = "2020-01-01"
END = "2026-01-01"

spx = yf.download(
    "^GSPC",
    start=START,
    end=END,
    auto_adjust=False
)

spx = spx.reset_index()

spx = spx[
    ["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"]
]

# Feature engineering
# rt​=ln(Pt​/Pt−1​)
spx["SPX_return"] = np.log(
    spx["Adj Close"] / spx["Adj Close"].shift(1)
)

# Rolling volatility 5 days
spx["SPX_volatility_5"] = (
    spx["SPX_return"]
    .rolling(5)
    .std()
)

# 20 days
spx["SPX_volatility_20"] = (
    spx["SPX_return"]
    .rolling(20)
    .std()
)


spx_final = spx[
    [
        "Date",
        "SPX_return",
        "SPX_volatility_5",
        "SPX_volatility_20"
    ]
]
```

```python
spx_final.to_csv("SPX500_features.csv", index=False)

from google.colab import files
files.download("SPX500_features.csv")
```
